```python
import cv2
import matplotlib.pyplot as plt
import os 
import numpy as np

current_view = cv2.imread("Drone Image.jpg")
plt.imshow(current_view)

current_view_hsv = cv2.cvtColor(current_view, cv2.COLOR_BGR2HSV)
h,s,v = cv2.split(current_view_hsv)
plt.figure(figsize=(12,4))
plt.subplot(131), plt.imshow(h, cmap='gray'), plt.title('Hue Channel')
plt.subplot(132), plt.imshow(s, cmap='gray'), plt.title('Saturation Channel')
plt.subplot(133), plt.imshow(v, cmap='gray'), plt.title('Value Channel')
plt.show()

current_view_gray = cv2.cvtColor(current_view, cv2.COLOR_BGR2GRAY)
retval, img_thresh = cv2.threshold(current_view_gray, 90, 180, cv2.THRESH_BINARY)
img_thresh_adp = cv2.adaptiveThreshold(current_view_gray, 255, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, 11, 7)
th,thresOtsu_img = cv2.threshold(current_view_gray,200,255,cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
plt.figure(figsize=(12,4))
plt.subplot(141),plt.imshow(current_view_gray, cmap='gray')
plt.subplot(142),plt.imshow(img_thresh, cmap='gray')
plt.subplot(143),plt.imshow(img_thresh_adp, cmap='gray')
plt.subplot(144),plt.imshow(thresOtsu_img, cmap='gray')
```
